# 什么是 LangChain

LangChain 是一个用于开发由大型语言模型（LLM）驱动的应用程序的开源框架。
它本身不开发 LLM，而是为各种 LLM 提供通用的接口，把 LLM 相关的组件“链接”在一起，简化 LLM 应用的开发难度。

简单来说，如果把大语言模型比作一颗性能惊人的“智能内核”，LangChain 就是围绕它建立起来的工程骨架。
它让开发者像搭积木一样，将各种组件连接起来，构建完整的 AI 应用链路。

LangChain 的核心价值在于模块化设计——将复杂任务拆解为可复用的组件，其强大之处正是来自各个组件如何协同工作以创建复杂的 AI 应用程序。

各个组件的具体说明：
|组件|用途|关键说明|
|----|-------|---------------|
|模型(Models)|AI 推理与生成|聊天模型、LLM、嵌入模型，负责文本生成与语义理解|
|提示 (Prompts)|结构化输入引导|指导语言模型如何响应的文本模板|
|链 (Chains)|多步骤工作流|将多个操作按顺序连接，上一步的输出作为下一步的输入|
|智能体 (Agents)|动态决策与编排|决定模型采取哪些行动，执行并观察流程直到完成为止|
|记忆 (Memory)|上下文保持|标准 LLM 是无状态的，记忆组件让 AI 能在多轮对话中保持上下文|
|检索器 (Retrievers)|信息访问|从外部数据源（文档、数据库）查找相关信息|
|工具 (Tools)|外部能力扩展|API、数据库、网络搜索等外部功能|

# 第一个 Agent 实例

```python
# 1. 导入 create_agent 函数
from langchain.agents import create_agent
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI

from settings import settings


# 2. 定义一个简单的工具函数
@tool
def get_weather(city: str) -> str:
    """获取指定城市的天气"""
    return f"{city}的天气总是晴朗的！"

# 3 创建 model 和 Agent
model = ChatOpenAI(
    model="deepseek-v4-flash",              # 也可以换用其他模型
    base_url="https://api.deepseek.com",    # deepseek api路径兼容 OpenAI
    api_key=settings.deepseek_api_key
)

agent = create_agent(
    model=model,  # 指定使用的模型
    tools=[get_weather],     # 注册工具
    system_prompt="你是一个乐于助人的助手。", # 系统提示词
)

# 4. 运行 Agent
result = agent.invoke(
    {"messages": [{"role": "user", "content": "天津的天气怎么样？"}]}
)

# 5. 打印结果
print(result["messages"][-1].content_blocks)
```
